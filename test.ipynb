```python
from pathlib import Path

import pandas as pd


PROJECT_ROOT = Path.home() / "Projects" / "OSN_pushback_detection"
TEST_DATA_ROOT = PROJECT_ROOT / "data" / "test" / "01_fetched"


def season_from_month(month: int) -> str:
    if month in (12, 1, 2):
        return "Winter"
    if month in (3, 4, 5):
        return "Spring"
    if month in (6, 7, 8):
        return "Summer"
    return "Autumn"


rows = []

for path in sorted(TEST_DATA_ROOT.rglob("*.parquet")):

    relative = path.relative_to(TEST_DATA_ROOT)

    airport = relative.parts[0]
    folder_season = relative.parts[1]
    filename = path.stem

    data = pd.read_parquet(
        path,
        columns=["timestamp"],
    )

    timestamps = pd.to_datetime(
        data["timestamp"],
        utc=True,
        errors="coerce",
    ).dropna()

    if timestamps.empty:
        continue

    start = timestamps.min()
    end = timestamps.max()

    actual_date = start.date()
    actual_season = season_from_month(start.month)

    try:
        filename_date = pd.to_datetime(
            filename,
            format="%d-%m-%y",
        ).date()
    except ValueError:
        filename_date = None

    rows.append(
        {
            "airport": airport,
            "folder_season": folder_season,
            "filename": filename,
            "filename_date": filename_date,
            "actual_date": actual_date,
            "actual_season": actual_season,
            "start": start,
            "end": end,
            "date_matches_filename": filename_date == actual_date,
            "season_matches_folder": folder_season == actual_season,
        }
    )


check = pd.DataFrame(rows)

check
```
